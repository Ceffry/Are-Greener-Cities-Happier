## Creating a master sheet

Keeping the year 2019, since it's the closest to the green space data (2018)

```python
quality_life_cl = quality_life_cl[quality_life_cl["year"] == 2019].copy()
```

```python
quality_life_cl["city"].unique()
```

Total matching cities: 29

Keep the positive survey results for city and green space satisfaction

```python
# City satisfaction: strongly agree
city_strong = quality_life_cl[quality_life_cl["id"] == "PS3360V"][["city", "year", "percentage"]].copy()

city_strong = city_strong.rename(columns={"percentage": "city_strongly_agree"})

# City satisfaction: somewhat agree
city_somewhat = quality_life_cl[quality_life_cl["id"] == "PS3361V"][["city", "year", "percentage"]].copy()

city_somewhat = city_somewhat.rename(columns={"percentage": "city_somewhat_agree"})
```

```python
# Green spaces: very satisfied
green_very = quality_life_cl[quality_life_cl["id"] == "PS1052V"][["city", "year", "percentage"]].copy()

green_very = green_very.rename(columns={"percentage": "green_very_satisfied"})

# Green spaces: rather satisfied
green_rather = quality_life_cl[quality_life_cl["id"] == "PS1053V"][["city", "year", "percentage"]].copy()

green_rather = green_rather.rename(columns={"percentage": "green_rather_satisfied"})
```

merge the tables into a master sheet

```python
survey_master = city_strong.merge(city_somewhat, 
                                  on=["city", "year"], 
                                  how="inner")

survey_master = survey_master.merge(green_very, 
                                    on=["city", "year"], 
                                    how="inner")

survey_master = survey_master.merge(green_rather, 
                                    on=["city", "year"], 
                                    how="inner")

survey_master
```

```python
survey_master = survey_master.merge(green_space_df,
                                    on="city",
                                    how="inner")
survey_master
```

Drop year column, since it's not going to be used for the analysis

```python
survey_master = survey_master.drop(columns="year")

survey_master
```

```python
# adding countries to the master sheet
city_to_country = {
    "Vienna": "Austria",
    "Brussels": "Belgium",
    "Sofia": "Bulgaria",
    "Nicosia": "Cyprus",
    "Prague": "Czechia",
    "Berlin": "Germany",
    "Copenhagen": "Denmark",
    "Tallinn": "Estonia",
    "Athens": "Greece",
    "Madrid": "Spain",
    "Paris": "France",
    "Zagreb": "Croatia",
    "Budapest": "Hungary",
    "Dublin": "Ireland",
    "Reykjavík": "Iceland",
    "Rome": "Italy",
    "Vilnius": "Lithuania",
    "Luxembourg": "Luxembourg",
    "Riga": "Latvia",
    "Valletta": "Malta",
    "Amsterdam": "Netherlands",
    "Oslo": "Norway",
    "Warsaw": "Poland",
    "Lisbon": "Portugal",
    "Bucharest": "Romania",
    "Stockholm": "Sweden",
    "Ljubljana": "Slovenia",
    "Bratislava": "Slovakia",
    "Ankara": "Türkiye"
}
survey_master["country"] = survey_master["city"].map(city_to_country)
```
